# MACHINE LEARNING LAB — TASK 5
## Cardio Disease Prediction — Classification

**Model:** Logistic Regression  
**Dataset:** `cardio_train.csv`

This notebook follows the Task 5 checklist for a classification problem:
- Accuracy, Precision, Recall and F1-score
- Train vs Test score
- Overfitting / Underfitting check
- 5-Fold Cross-Validation
- Model comparison
- Hyperparameter tuning using GridSearchCV
- Final model evaluation


## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

import pickle
```

## 2. Load the Cardio Dataset

```python
df = pd.read_csv("cardio_train.csv", sep=";")

print("Dataset shape:", df.shape)
df.head()
```

## 3. Basic Dataset Information

```python
print("Columns:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

## 4. Remove Duplicate Rows

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

## 5. Handle Outliers

For the main continuous measurements, the IQR method is used for:
- `height`
- `weight`
- `ap_hi`
- `ap_lo`


```python
outlier_columns = ["height", "weight", "ap_hi", "ap_lo"]

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    before = len(df)
    df = df[(df[col] >= lower) & (df[col] <= upper)]
    after = len(df)

    print(col, "removed:", before - after)

print("Shape after IQR outlier handling:", df.shape)
```

## 6. Select Features (X) and Target (y)

The target is `cardio`.

The `id` column is removed because it is only an identifier and is not a useful prediction feature.


```python
X = df.drop(["cardio", "id"], axis=1)
y = df["cardio"]

print("Features:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

## 7. Split Data into Training and Testing Sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

## 8. Scale the Features

`fit_transform()` is used only on training data.

`transform()` is used on test data because the scaler must use the parameters learned from the training data.


```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

## 9. Train Logistic Regression Model

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully.")
```

## 10. Make Predictions

```python
y_train_pred = model.predict(X_train_scaled)
y_test_pred = model.predict(X_test_scaled)

print("Predictions completed.")
```

## 11. Calculate Classification Metrics

Required Task 5 metrics:
- Accuracy
- Precision
- Recall
- F1-score


```python
train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_test_pred)

precision = precision_score(y_test, y_test_pred)
recall = recall_score(y_test, y_test_pred)
f1 = f1_score(y_test, y_test_pred)

print("Training Accuracy :", round(train_accuracy, 4))
print("Testing Accuracy  :", round(test_accuracy, 4))
print("Precision         :", round(precision, 4))
print("Recall            :", round(recall, 4))
print("F1-score          :", round(f1, 4))
```

## 12. Classification Report

```python
print(classification_report(y_test, y_test_pred))
```

## 13. Confusion Matrix

```python
cm = confusion_matrix(y_test, y_test_pred)

print("Confusion Matrix:")
print(cm)

plt.figure(figsize=(5, 4))
plt.imshow(cm)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.colorbar()

for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.show()
```

## 14. Check Overfitting / Underfitting

Compare training and testing accuracy.

- Training much higher than testing → possible overfitting
- Both scores low → possible underfitting
- Scores close → good fit


```python
difference = train_accuracy - test_accuracy

print("Training Accuracy:", round(train_accuracy, 4))
print("Testing Accuracy :", round(test_accuracy, 4))
print("Difference        :", round(difference, 4))

if difference > 0.05:
    print("Result: Possible overfitting.")
elif train_accuracy < 0.60 and test_accuracy < 0.60:
    print("Result: Possible underfitting.")
else:
    print("Result: Training and testing scores are reasonably close.")
```

## 15. 5-Fold Cross-Validation

Cross-validation checks whether the model is stable across different parts of the training data.


```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_scores = cross_val_score(
    model,
    X_train_scaled,
    y_train,
    cv=cv,
    scoring="accuracy"
)

print("5-Fold Accuracy Scores:")
for i, score in enumerate(cv_scores, start=1):
    print(f"Fold {i}: {score:.4f}")

print("\nAverage CV Accuracy:", round(cv_scores.mean(), 4))
print("CV Score Spread    :", round(cv_scores.max() - cv_scores.min(), 4))
```

## 16. Visualize Cross-Validation Scores

```python
plt.figure(figsize=(7, 4))
plt.plot(range(1, 6), cv_scores, marker="o")
plt.xlabel("Fold")
plt.ylabel("Accuracy")
plt.title("5-Fold Cross-Validation Accuracy")
plt.xticks(range(1, 6))
plt.grid(True)
plt.show()
```

## 17. Model Comparison Table

Only Logistic Regression is implemented in this project, so the comparison table records the model actually trained rather than creating unsupported results for other models.


```python
comparison = pd.DataFrame({
    "Model": ["Logistic Regression"],
    "Accuracy": [test_accuracy],
    "Precision": [precision],
    "Recall": [recall],
    "F1-score": [f1]
})

comparison
```

## 18. Hyperparameter Tuning with GridSearchCV

For Logistic Regression, we tune:
- `C`
- `solver`


```python
param_grid = {
    "C": [0.01, 0.1, 1, 10, 100],
    "solver": ["lbfgs", "liblinear"]
}

grid_search = GridSearchCV(
    LogisticRegression(max_iter=1000),
    param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid_search.fit(X_train_scaled, y_train)

print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest CV Accuracy:")
print(round(grid_search.best_score_, 4))
```

## 19. Test the Tuned Model

```python
tuned_model = grid_search.best_estimator_

y_tuned_pred = tuned_model.predict(X_test_scaled)

tuned_accuracy = accuracy_score(y_test, y_tuned_pred)
tuned_precision = precision_score(y_test, y_tuned_pred)
tuned_recall = recall_score(y_test, y_tuned_pred)
tuned_f1 = f1_score(y_test, y_tuned_pred)

print("Tuned Test Accuracy :", round(tuned_accuracy, 4))
print("Tuned Precision     :", round(tuned_precision, 4))
print("Tuned Recall        :", round(tuned_recall, 4))
print("Tuned F1-score      :", round(tuned_f1, 4))

print("\nAccuracy improvement:", round(tuned_accuracy - test_accuracy, 4))
```

## 20. Final Model Comparison

```python
final_comparison = pd.DataFrame({
    "Version": ["Before Tuning", "After Tuning"],
    "Accuracy": [test_accuracy, tuned_accuracy],
    "Precision": [precision, tuned_precision],
    "Recall": [recall, tuned_recall],
    "F1-score": [f1, tuned_f1]
})

final_comparison
```

## 21. Save the Model and Scaler as PKL Files

These files can be used later for prediction without retraining the model.


```python
# Save the tuned Logistic Regression model
with open("cardio_model.pkl", "wb") as file:
    pickle.dump(tuned_model, file)

# Save the scaler
with open("cardio_scaler.pkl", "wb") as file:
    pickle.dump(scaler, file)

print("cardio_model.pkl and cardio_scaler.pkl saved successfully.")
```

## 22. Load PKL Files and Test a New Patient

The feature order must remain exactly the same as the training data.


```python
with open("cardio_model.pkl", "rb") as file:
    loaded_model = pickle.load(file)

with open("cardio_scaler.pkl", "rb") as file:
    loaded_scaler = pickle.load(file)

# Example patient:
# age, gender, height, weight, ap_hi, ap_lo,
# cholesterol, gluc, smoke, alco, active
new_patient = [[55, 2, 170, 70, 120, 80, 1, 1, 0, 0, 1]]

new_patient_scaled = loaded_scaler.transform(new_patient)
prediction = loaded_model.predict(new_patient_scaled)

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Result: Higher risk of cardiovascular disease.")
else:
    print("Result: Lower risk of cardiovascular disease.")
```
